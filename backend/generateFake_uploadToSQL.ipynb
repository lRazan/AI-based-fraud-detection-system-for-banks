```python
import pandas as pd
from faker import Faker

fake = Faker()

#read file cvs 
df = pd.read_csv('/Users/razan/Documents/Master/AI/transactions_with_id.csv')

# Data extraction based on nameOrig(beacse may the one custore do multiple of transaction)
unique_customers = df['nameOrig'].unique()


customers_data = []

for customer_id in unique_customers:
    name = fake.name()
    phone_number = fake.phone_number()
    city = fake.city()
    customers_data.append((customer_id, name, phone_number, city))

# new DataFrame
customers_df = pd.DataFrame(customers_data, columns=['customer_id', 'name', 'phone_number', 'city'])

# to CSV
customers_df.to_csv('correct_customers.csv', index=False)

print("corret upload correct_customers.csv")
```

```python
import pandas as pd
import mysql.connector

# connect to MySQL
conn = mysql.connector.connect(
    host='localhost',
    user='root',
    password='User0202',   # 🔵 password
    database='fraud_detection'
)
cursor = conn.cursor()

# read file
df = pd.read_csv('/Users/razan/Documents/Master/AI/correct_customers.csv')

# uplaod file as batches 
batch_size = 5000

for start in range(0, len(df), batch_size):
    end = start + batch_size
    batch = df.iloc[start:end]
    records = batch.values.tolist()

    query = """
    INSERT INTO customers (customer_id, name, phone_number, city)
    VALUES (%s, %s, %s, %s)
    """
    cursor.executemany(query, records)
    conn.commit()
    print(f"Inserted customers {start} to {end}")

cursor.close()
conn.close()

print("All correct customers inserted successfully!")
```
